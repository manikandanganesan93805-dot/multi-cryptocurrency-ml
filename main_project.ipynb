```python
import pandas as pd
import numpy as np

from sklearn.ensemble import RandomForestClassifier

from sklearn.model_selection import train_test_split

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)
```

```python
from google.colab import files

uploaded = files.upload()

file_name = next(iter(uploaded))
df = pd.read_csv(file_name)

print("Dataset loaded successfully!")
print("Number of Rows    :", df.shape[0])
print("Number of Columns :", df.shape[1])

display(df.head())
```

```python
print(df.info())

print(df.shape)

df = df.drop_duplicates()

print(df["Date"].isnull().sum())

df = df.dropna(subset=["Date"])
df = df.sort_values(
    by=["Coin", "Date"]
).reset_index(drop=True)

print("Rows    :", df.shape[0])
print("Columns :", df.shape[1])
print(df.isnull().sum())

display(df.head())
```

```python
print("Number of Different Coins:",
      df["Coin"].nunique())

print("\nCoin Names:")
print(df["Coin"].unique())

print("\nRecords Per Coin:")
print(df["Coin"].value_counts())
```

```python
print("Final Dataset Shape:", df.shape)

print("\nTotal Missing Values:",
      df.isnull().sum().sum())

print("Total Duplicate Rows:",
      df.duplicated().sum())

display(df.head(10))
```

```python
print("Dataset Shape:")
print(df.shape)

print("\nDataset Columns:")
print(df.columns.tolist())

print("\nStatistical Summary:")
display(df.describe())

coin_counts = df["Coin"].value_counts()

print("Number of Records for Each Coin:")
display(coin_counts)
```

```python

df = df.sort_values(
    ["Coin", "Date"]
).reset_index(drop=True)

df["Return_1D"] = (
    df.groupby("Coin")["Close"]
      .pct_change()
)

print("Return_1D created successfully!")

display(
    df[
        ["Date", "Coin", "Close", "Return_1D"]
    ].head(10)
)
```

```python
df["MA_7"] = (
    df.groupby("Coin")["Close"]
      .transform(
          lambda x: x.rolling(
              window=7,
              min_periods=1
          ).mean()
      )
)

print("MA_7 created successfully!")

display(
    df[
        ["Date", "Coin", "Close", "MA_7"]
    ].head(10)
)
```

```python
df["MA_14"] = (
    df.groupby("Coin")["Close"]
      .transform(
          lambda x: x.rolling(
              window=14,
              min_periods=1
          ).mean()
      )
)

print("MA_14 created successfully!")

display(
    df[
        ["Date", "Coin", "Close", "MA_14"]
    ].head(10)
)
```

```python
df["Volatility_7D"] = (
    df.groupby("Coin")["Return_1D"]
      .transform(
          lambda x: x.rolling(
              window=7,
              min_periods=2
          ).std()
      )
)

print("Volatility_7D created successfully!")

display(
    df[
        ["Date", "Coin", "Return_1D", "Volatility_7D"]
    ].head(10)
)
```

```python
df["Future_Return_1D"] = (
    df.groupby("Coin")["Close"]
      .shift(-1) / df["Close"] - 1
)

print("Future_Return_1D created successfully!")

display(
    df[
        [
            "Date",
            "Coin",
            "Close",
            "Future_Return_1D"
        ]
    ].head(10)
)
```

```python
df["Target"] = np.where(
    df["Future_Return_1D"] > 0,
    1,
    0
)

# Last row of each coin has no future value
# so mark it as missing
last_rows = (
    df.groupby("Coin")
      .tail(1)
      .index
)

df.loc[last_rows, "Target"] = np.nan

print("Target created successfully!")

display(
    df[
        [
            "Date",
            "Coin",
            "Close",
            "Future_Return_1D",
            "Target"
        ]
    ].head(15)
)
```

```python
print(
    df["Target"]
    .value_counts(dropna=False)
)
target_by_coin = pd.crosstab(
    df["Coin"],
    df["Target"]
)

print("UP/DOWN Distribution by Coin:")

display(target_by_coin)
```

```python
features = [
    "Open",
    "High",
    "Low",
    "Close",
    "Volume",
    "Return_1D",
    "MA_7",
    "MA_14",
    "Volatility_7D"
]

target = "Target"

print("ML Features:")
for feature in features:
    print("-", feature)

print("\nTarget:")
print("-", target)
```

```python
model_df = df.copy()

model_df = model_df.dropna(
    subset=features + ["Target"]
)

model_df["Target"] = (
    model_df["Target"]
    .astype(int)
)

print("Final ML Dataset Shape:")
print(model_df.shape)

print("\nMissing Values:")
print(model_df.isnull().sum())

display(model_df.head())
```

```python
train_parts = []
test_parts = []

# Split each cryptocurrency separately
for coin, group in model_df.groupby("Coin"):

    # Sort by date
    group = group.sort_values("Date").reset_index(drop=True)

    # Calculate 80% split point
    split_index = int(len(group) * 0.80)

    # First 80% = Training
    train_coin = group.iloc[:split_index]

    # Last 20% = Testing
    test_coin = group.iloc[split_index:]

    train_parts.append(train_coin)
    test_parts.append(test_coin)

# Combine all coins
train = pd.concat(
    train_parts,
    ignore_index=True
)

test = pd.concat(
    test_parts,
    ignore_index=True
)

X_train = train[features]
y_train = train["Target"]

X_test = test[features]
y_test = test["Target"]

print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("X_test :", X_test.shape)
print("y_test :", y_test.shape)

print("Start Date:", train["Date"].min())
print("End Date  :", train["Date"].max())

print("Start Date:", test["Date"].min())
print("End Date  :", test["Date"].max())

print("Train Dataset Shape:", train.shape)
print("Test Dataset Shape :", test.shape)
```

```python
for coin in model_df["Coin"].unique():

    train_count = len(
        train[train["Coin"] == coin]
    )

    test_count = len(
        test[test["Coin"] == coin]
    )

    print(
        f"{coin:10s} -> "
        f"Train: {train_count:5d} | "
        f"Test: {test_count:5d}"
    )
print(y_train.value_counts())
print(y_test.value_counts())
```

```python
model = RandomForestClassifier(
    n_estimators=100,
    max_depth=10,
    random_state=42,
    n_jobs=-1
)

model.fit(X_train, y_train)

print("Training completed!")
y_pred = model.predict(X_test)

print("Prediction completed!")

accuracy = accuracy_score(y_test, y_pred)

print(f"Accuracy: {accuracy * 100:.2f}%")

comparison = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred
})

print(
    classification_report( y_test,y_pred,target_names=["DOWN", "UP"]
    )
)
```

```python
import gradio as gr

coins = sorted(
    model_df["Coin"].unique()
)

print("Available Coins:")
print(coins)

latest_data = (
    model_df
    .sort_values(["Coin", "Date"])
    .groupby("Coin")
    .tail(1)
    .copy()
)

def predict_coin(coin_name):

    coin_data = latest_data[
        latest_data["Coin"] == coin_name
    ]

    if coin_data.empty:
        return (
            "No Data",
            "0.00%",
            "0.00%",
            "Random Forest",
            f"{accuracy * 100:.2f}%"
        )

    # Use the SAME features used during training
    X_input = coin_data[features].copy()

    # Clean input
    X_input = X_input.replace(
        [np.inf, -np.inf],
        np.nan
    ).fillna(0)

    # Random Forest prediction
    prediction = model.predict(X_input)[0]

    # Probability
    probability = model.predict_proba(X_input)[0]

    down_probability = probability[0] * 100
    up_probability = probability[1] * 100

    # UP / DOWN
    if prediction == 1:
        movement = "🟢 UP"
    else:
        movement = "🔴 DOWN"

    return (
        movement,
        f"{up_probability:.2f}%",
        f"{down_probability:.2f}%",
        "Random Forest",
        f"{accuracy * 100:.2f}%"
    )



with gr.Blocks(
    title="Cryptocurrency Prediction"
) as app:

    gr.Markdown(
        """
        #  Cryptocurrency Price Movement Prediction

        ### Random Forest Machine Learning Model
        """
    )

    coin_dropdown = gr.Dropdown(
        choices=coins,
        value=coins[0],
        label=" Select Cryptocurrency"
    )

    predict_button = gr.Button(
        " Predict UP / DOWN",
        variant="primary"
    )

    # Results in pridiction
    movement_output = gr.Textbox(
        label=" Predicted Movement"
    )

    up_output = gr.Textbox(
        label="🟢 UP Probability"
    )

    down_output = gr.Textbox(
        label="🔴 DOWN Probability"
    )

    model_output = gr.Textbox(
        label=" Model Used"
    )

    accuracy_output = gr.Textbox(
        label="Random Forest Test Accuracy"
    )

    predict_button.click(
        fn=predict_coin,
        inputs=coin_dropdown,
        outputs=[
            movement_output,
            up_output,
            down_output,
            model_output,
            accuracy_output
        ]
    )



app.launch(
    share=True
)
```
